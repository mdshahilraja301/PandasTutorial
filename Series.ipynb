Pandas Series

```python
import numpy as np
import pandas as pd
```

```python
labels = ['A','B','C','D']
my_list = [1,2,3,4]
arr = np.array([3,4,5,6])
d = {'a':1,'b':2,'c':3,'d':4}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list,index = labels)
```

```python
pd.Series(arr,index = labels)      #only 1D array is accepet here
```

```python
pd.Series(d)
```
