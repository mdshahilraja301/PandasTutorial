Create DataFrame

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
```

```python
pd.DataFrame(data_list)
```

```python
columns=['Name', 'Age', 'City', 'Salary']
df2 = pd.DataFrame(data_list,columns = columns)
df2
```

# **Selection and indexing of column**

```python
df2['Name']
```

```python
df2[['Name','Age']]
```

# **Create New column**

```python
df2['Designation'] = ['Doctor','Engneer','DS Eng.','AI Eng.']
df2
```

# **Remove A Column and Row**

```python
df2 = df2.drop('Designation',axis = 1)
df2
```

```python
df2.drop(0)       #for permanent delete   'df2.drop(0,inplace=True)'
```

```python
#I only want to see those people whose age is above 30
df2[df2['Age'] > 30]
```
