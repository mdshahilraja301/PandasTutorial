```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
```

**Finding Missing Data**

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

**Removing Missing Data**

```python
df.dropna()        #Remove all Rows which Row have null value
```

```python
df.dropna(thresh = 3)    #Remove all Rows which Row have less than 3 null vale
```

**Filling Missing Data**

```python
df.fillna(value = 0)
```

```python
values = {'A': 10, 'B': 20, 'C': 30, 'D': 40}
df.fillna(value = values)
```

```python
df.fillna(value = df.mean())
```
